```python
# DFA to recognize the regular expression
# (0+1)*(00+11)
#the language contains all binary strings that end with either 00 or 11.

#Examples accepted: 00, 11, 000, 011, 100, 111, 0100
#Examples rejected: 0, 1, 01, 10, 001, 110,0101

def dfa(string):
    # Initial state
    state = 'q0'

    for symbol in string:

        if symbol not in ['0', '1']:
            return False

        # DFA transition table
        if state == 'q0':
            if symbol == '0':
                state = 'q1'
            else:
                state = 'q2'

        elif state == 'q1':
            if symbol == '0':
                state = 'q3'
            else:
                state = 'q2'

        elif state == 'q2':
            if symbol == '0':
                state = 'q1'
            else:
                state = 'q4'

        elif state == 'q3':
            if symbol == '0':
                state = 'q3'
            else:
                state = 'q2'

        elif state == 'q4':
            if symbol == '0':
                state = 'q1'
            else:
                state = 'q4'

    # q3 and q4 are final/accepting states
    return state in ['q3', 'q4']


# Main program
word = input("Enter a word: ")

if dfa(word):
    print("String Accepted")
    print("The word is recognized by the DFA.")
else:
    print("String Rejected")
    print("The word is not recognized by the DFA.")
```
