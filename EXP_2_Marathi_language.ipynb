```python
import nltk
import stanza
from collections import Counter
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
stanza.download('mr')
```

```python
nlp_mr = stanza.Pipeline(
    lang='mr',
    processors='tokenize,pos,lemma'
)
```

Marathi input text

```python
marathi_text = """
मराठी ही भारतातील एक प्रमुख भाषा आहे.
महाराष्ट्र राज्यात मराठी भाषा मोठ्या प्रमाणावर बोलली जाते.
मराठी साहित्याला समृद्ध आणि दीर्घ परंपरा आहे.
विद्यार्थी दररोज नवीन गोष्टी शिकतात.
तंत्रज्ञानामुळे शिक्षण अधिक सोपे झाले आहे.
"""
```

Sentence Detection using Stanza

```python
doc = nlp_mr(marathi_text)

print("Marathi Sentences:\n")

for i, sentence in enumerate(doc.sentences, 1):
    print(f"Sentence {i}: {sentence.text}")
```

Tokenization using Stanza

```python
print("Marathi Tokens:\n")

tokens = []

for sentence in doc.sentences:
    for word in sentence.words:
        tokens.append(word.text)

print(tokens)
```

Tokenization with Statistical Analysis



```python
from collections import Counter

# Tokenization
tokens = []

for sentence in doc.sentences:
    for word in sentence.words:
        tokens.append(word.text)

print("===== TOKENIZATION =====")
print(tokens)

# Statistical Analysis
total_tokens = len(tokens)
unique_tokens = len(set(tokens))

# Average token length
avg_token_length = sum(len(token) for token in tokens) / total_tokens

# Top 5 most frequent tokens
token_frequency = Counter(tokens)
top_5_tokens = token_frequency.most_common(5)

print("\n===== STATISTICAL ANALYSIS =====")
print("Total Tokens:", total_tokens)
print("Unique Tokens:", unique_tokens)
print("Average Token Length:", round(avg_token_length, 2))

print("\nTop 5 Most Frequent Tokens:")
for token, frequency in top_5_tokens:
    print(token, ":", frequency)
```

Lemmatization using Stanza

```python
print("Marathi Lemmatization:\n")

for sentence in doc.sentences:
    for word in sentence.words:
        print(word.text, "->", word.lemma)
```

Stop Word Removal

```python
# Small custom Marathi stop word list, since Stanza has no built-in list
marathi_stopwords = {
    'आहे', 'ही', 'हा', 'हे', 'एक', 'आणि', 'च्या',
    'मध्ये', 'वर', 'पासून', 'साठी', 'की', 'तो',
    'ती', 'ते', 'मी', 'तू', 'आपण'
}
filtered_tokens = [
    token for token in tokens
    if token not in marathi_stopwords
]

print("Tokens after Stop Word Removal:\n")
print(filtered_tokens)


```

```python
marathi_text = """
मराठी ही भारतातील एक प्रमुख भाषा आहे.
महाराष्ट्र राज्यात मराठी भाषा मोठ्या प्रमाणावर बोलली जाते.
मराठी साहित्याला समृद्ध आणि दीर्घ परंपरा आहे.
विद्यार्थी दररोज नवीन गोष्टी शिकतात.
तंत्रज्ञानामुळे शिक्षण अधिक सोपे झाले आहे."""
```

Sentence Detection using NLTK


```python
import nltk
sentences = nltk.tokenize.RegexpTokenizer(r'[^.]+').tokenize(marathi_text)

print("===== SENTENCE DETECTION =====")

for i, sentence in enumerate(sentences, 1):
    print(f"Sentence {i}: {sentence.strip()}")
```

Tokenization using NLTK

```python
tokens = []

for sentence in sentences:
    sentence_tokens = nltk.word_tokenize(sentence)
    tokens.extend(sentence_tokens)

print("===== TOKENIZATION =====")
print(tokens)
```

Statistical Analysis using NLTK

```python
from collections import Counter

total_tokens = len(tokens)

unique_tokens = len(set(tokens))

avg_token_length = sum(
    len(token) for token in tokens
) / total_tokens

token_frequency = Counter(tokens)

top_5_tokens = token_frequency.most_common(5)

print("===== STATISTICAL ANALYSIS OVER TOKENS =====")

print("Total Tokens:", total_tokens)

print("Unique Tokens:", unique_tokens)

print("Average Token Length:",
      round(avg_token_length, 2))

print("\nTop 5 Most Frequent Tokens:")

for token, frequency in top_5_tokens:
    print(f"{token} : {frequency}")
```

 Stop Word Removal using NLTK

NLTK does not provide Marathi stopwords directly, so create a Marathi stopword set





```python
marathi_stopwords = {
    'आहे', 'ही', 'हा', 'हे', 'एक',
    'आणि', 'च्या', 'मध्ये', 'वर',
    'पासून', 'साठी', 'की', 'तो',
    'ती', 'ते', 'मी', 'तू', 'आपण'
}

filtered_tokens = [
    token for token in tokens
    if token not in marathi_stopwords
]

print("===== STOP WORD REMOVAL =====")
print(filtered_tokens)
```
