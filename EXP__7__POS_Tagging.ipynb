```python

import nltk

# Download the required NLTK resources
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("averaged_perceptron_tagger")
nltk.download("averaged_perceptron_tagger_eng")

def pos_tag_sentence(sentence):
    # Tokenize the sentence
    tokens = nltk.word_tokenize(sentence)

    # Perform POS tagging
    tagged_words = nltk.pos_tag(tokens)

    return tagged_words


# Multiple input sentences for exploration
sentences = [
    "Book that flight.",
    "Hand me that book.",
    "Does that flight serve dinner?",
    "I read a book.",
    "Please book a ticket.",
    "The quick brown fox jumps over the lazy dog.",
    "She is running in the park.",
    "The children are playing football.",
    "He can swim.",
    "I can a can.",
    "Time flies like an arrow.",
    "Fruit flies like a banana.",
    "The students study NLP.",
    "The beautiful flowers bloom in spring.",
    "She writes a letter and he reads a book."
]

# Display POS tags for every sentence
for i, sentence in enumerate(sentences, start=1):
    print(f"\nInput {i}: {sentence}")
    print("Output:")

    tagged_words = pos_tag_sentence(sentence)

    for word, tag in tagged_words:
        print(f"{word} / {tag}", end="   ")

    print()
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")

sentences = [
    "Book that flight.",
    "Hand me that book.",
    "Does that flight serve dinner?",
    "I read a book.",
    "Please book a ticket.",
    "The quick brown fox jumps over the lazy dog.",
    "She is running in the park.",
    "The children are playing football.",
    "He can swim.",
    "The students study NLP.",
    "The beautiful flowers bloom in spring.",
    "She writes a letter and he reads a book.",
    "Time flies like an arrow.",
    "Fruit flies like a banana.",
    "Python is a powerful programming language."
]

for sentence in sentences:
    doc = nlp(sentence)

    print("\nInput:", sentence)
    print("Output:", end=" ")

    for token in doc:
        print(f"{token.text} / {token.tag_}", end="   ")

    print()
```

```python
import stanza

# Download the English model
stanza.download("en")

# Initialize the English pipeline
nlp = stanza.Pipeline(
    lang="en",
    processors="tokenize,pos",
    verbose=False
)

sentences = [
    "Book that flight.",
    "Hand me that book.",
    "Does that flight serve dinner?",
    "I read a book.",
    "Please book a ticket.",
    "The quick brown fox jumps over the lazy dog.",
    "She is running in the park.",
    "The children are playing football.",
    "He can swim.",
    "The students study NLP.",
    "The beautiful flowers bloom in spring.",
    "She writes a letter and he reads a book.",
    "Time flies like an arrow.",
    "Fruit flies like a banana.",
    "Python is a powerful programming language."
]

for sentence in sentences:
    doc = nlp(sentence)

    print("\nInput:", sentence)
    print("Output:", end=" ")

    for sent in doc.sentences:
        for word in sent.words:
            print(f"{word.text} / {word.xpos}", end="   ")

    print()
```
