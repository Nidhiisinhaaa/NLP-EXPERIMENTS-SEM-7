NLTK — Bi-Gram, Tri-Gram & Four-Gram

```python
# Install NLTK if required
!pip install nltk

import nltk
from nltk.util import ngrams
from collections import Counter

# Download tokenizer
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python

# --------------------------------------------------
# 1. CORPUS - 12 SENTENCES
# --------------------------------------------------

corpus = [
    "I like to study AI",
    "I like to study NLP",
    "I like AI projects",
    "AI is a branch of computer science",
    "NLP is a subdomain of AI",
    "NLP is useful for language processing",
    "Students like to study AI",
    "Students like NLP projects",
    "AI projects are interesting",
    "NLP projects are useful",
    "I study computer science",
    "I like computer science"
]

# Convert sentences into words
tokens = []

for sentence in corpus:
    words = nltk.word_tokenize(sentence.lower())
    tokens.extend(words)

# --------------------------------------------------
# 2. UNIGRAM / WORD FREQUENCY TABLE
# --------------------------------------------------

unigram = Counter(tokens)

print("\n========== WORD FREQUENCY TABLE ==========")
print("Word\t\tFrequency")
print("------------------------------------------")

for word, freq in unigram.most_common():
    print(f"{word:<15}{freq}")

# --------------------------------------------------
# 3. BIGRAM
# --------------------------------------------------

bigram = list(ngrams(tokens, 2))
bigram_freq = Counter(bigram)

print("\n========== BIGRAM FREQUENCY TABLE ==========")
print("Bigram\t\t\tFrequency")
print("---------------------------------------------")

for bg, freq in bigram_freq.most_common():
    print(f"{' '.join(bg):<25}{freq}")

# --------------------------------------------------
# 4. TRIGRAM
# --------------------------------------------------

trigram = list(ngrams(tokens, 3))
trigram_freq = Counter(trigram)

print("\n========== TRIGRAM FREQUENCY TABLE ==========")
print("Trigram\t\t\t\tFrequency")
print("----------------------------------------------")

for tg, freq in trigram_freq.most_common():
    print(f"{' '.join(tg):<30}{freq}")

# --------------------------------------------------
# 5. FOUR-GRAM
# --------------------------------------------------

fourgram = list(ngrams(tokens, 4))
fourgram_freq = Counter(fourgram)

print("\n========== FOUR-GRAM FREQUENCY TABLE ==========")
print("Four-gram\t\t\t\tFrequency")
print("------------------------------------------------")

for fg, freq in fourgram_freq.most_common():
    print(f"{' '.join(fg):<35}{freq}")

# --------------------------------------------------
# 6. NEXT WORD PREDICTION
# --------------------------------------------------

def predict_next_word(context, ngram_freq, n):
    context = tuple(context.lower().split())

    candidates = {}

    for gram, freq in ngram_freq.items():

        if gram[:-1] == context:
            candidates[gram[-1]] = freq

    if candidates:
        predicted_word = max(candidates, key=candidates.get)
        return predicted_word

    return "No prediction found"


# --------------------------------------------------
# BIGRAM PREDICTION
# --------------------------------------------------

print("\n========== NEXT WORD PREDICTION ==========")

context = "i like"

# For bigram, use last word
bigram_context = context.split()[-1]

prediction = predict_next_word(
    bigram_context,
    bigram_freq,
    2
)

print("\nBigram:")
print("Input : I like")
print("Predicted next word :", prediction)


# --------------------------------------------------
# TRIGRAM PREDICTION
# --------------------------------------------------

trigram_context = "i like"

prediction = predict_next_word(
    trigram_context,
    trigram_freq,
    3
)

print("\nTrigram:")
print("Input : I like")
print("Predicted next word :", prediction)


# --------------------------------------------------
# FOUR-GRAM PREDICTION
# --------------------------------------------------

fourgram_context = "i like to"

prediction = predict_next_word(
    fourgram_context,
    fourgram_freq,
    4
)

print("\nFour-gram:")
print("Input : I like to")
print("Predicted next word :", prediction)
```

```python
import spacy
from collections import Counter

# Load spaCy English model
nlp = spacy.load("en_core_web_sm")


# --------------------------------------------------
# 1. CORPUS - 12 SENTENCES
# --------------------------------------------------

corpus = [
    "I like to study AI",
    "I like to study NLP",
    "I like AI projects",
    "AI is a branch of computer science",
    "NLP is a subdomain of AI",
    "NLP is useful for language processing",
    "Students like to study AI",
    "Students like NLP projects",
    "AI projects are interesting",
    "NLP projects are useful",
    "I study computer science",
    "I like computer science"
]


# --------------------------------------------------
# 2. TOKENIZATION USING SPACY
# --------------------------------------------------

tokens = []

for sentence in corpus:
    doc = nlp(sentence.lower())

    for token in doc:
        if not token.is_space:
            tokens.append(token.text)

print("\n========== TOKENS ==========")
print(tokens)


# --------------------------------------------------
# 3. UNIGRAM
# --------------------------------------------------

unigram = Counter(tokens)

print("\n========== WORD FREQUENCY TABLE ==========")
print("Word\t\tFrequency")
print("------------------------------------------")

for word, freq in unigram.most_common():
    print(f"{word:<15}{freq}")


# --------------------------------------------------
# FUNCTION TO CREATE N-GRAMS
# --------------------------------------------------

def make_ngrams(tokens, n):
    return [
        tuple(tokens[i:i+n])
        for i in range(len(tokens) - n + 1)
    ]


# --------------------------------------------------
# 4. BIGRAM
# --------------------------------------------------

bigram = make_ngrams(tokens, 2)
bigram_freq = Counter(bigram)

print("\n========== BIGRAM FREQUENCY TABLE ==========")
print("Bigram\t\t\tFrequency")
print("---------------------------------------------")

for bg, freq in bigram_freq.most_common():
    print(f"{' '.join(bg):<25}{freq}")


# --------------------------------------------------
# 5. TRIGRAM
# --------------------------------------------------

trigram = make_ngrams(tokens, 3)
trigram_freq = Counter(trigram)

print("\n========== TRIGRAM FREQUENCY TABLE ==========")
print("Trigram\t\t\t\tFrequency")
print("----------------------------------------------")

for tg, freq in trigram_freq.most_common():
    print(f"{' '.join(tg):<30}{freq}")


# --------------------------------------------------
# 6. FOUR-GRAM
# --------------------------------------------------

fourgram = make_ngrams(tokens, 4)
fourgram_freq = Counter(fourgram)

print("\n========== FOUR-GRAM FREQUENCY TABLE ==========")
print("Four-gram\t\t\t\tFrequency")
print("------------------------------------------------")

for fg, freq in fourgram_freq.most_common():
    print(f"{' '.join(fg):<35}{freq}")


# --------------------------------------------------
# 7. NEXT WORD PREDICTION
# --------------------------------------------------

def predict_next_word(context, ngram_freq):

    context = tuple(context.lower().split())

    candidates = {}

    for gram, freq in ngram_freq.items():

        if gram[:-1] == context:
            candidates[gram[-1]] = freq

    if candidates:
        return max(candidates, key=candidates.get)

    return "No prediction found"


# --------------------------------------------------
# 8. PREDICTION
# --------------------------------------------------

print("\n========== NEXT WORD PREDICTION ==========")


# BIGRAM
prediction = predict_next_word(
    "like",
    bigram_freq
)

print("\nBigram:")
print("Input : I like")
print("Predicted next word :", prediction)


# TRIGRAM
prediction = predict_next_word(
    "i like",
    trigram_freq
)

print("\nTrigram:")
print("Input : I like")
print("Predicted next word :", prediction)


# FOUR-GRAM
prediction = predict_next_word(
    "i like to",
    fourgram_freq
)

print("\nFour-gram:")
print("Input : I like to")
print("Predicted next word :", prediction)
```

```python
import stanza
from collections import Counter

# Download English model
stanza.download('en')

# Create Stanford/Stanza pipeline
nlp = stanza.Pipeline(
    lang='en',
    processors='tokenize'
)
```

```python

# --------------------------------------------------
# 1. CORPUS - 12 SENTENCES
# --------------------------------------------------

corpus = [
    "I like to study AI",
    "I like to study NLP",
    "I like AI projects",
    "AI is a branch of computer science",
    "NLP is a subdomain of AI",
    "NLP is useful for language processing",
    "Students like to study AI",
    "Students like NLP projects",
    "AI projects are interesting",
    "NLP projects are useful",
    "I study computer science",
    "I like computer science"
]


# --------------------------------------------------
# 2. TOKENIZATION USING STANFORD NLP
# --------------------------------------------------

tokens = []

for sentence in corpus:

    doc = nlp(sentence.lower())

    for sent in doc.sentences:
        for word in sent.words:
            tokens.append(word.text)


print("\n========== TOKENS ==========")
print(tokens)


# --------------------------------------------------
# 3. UNIGRAM
# --------------------------------------------------

unigram = Counter(tokens)

print("\n========== WORD FREQUENCY TABLE ==========")
print("Word\t\tFrequency")
print("------------------------------------------")

for word, freq in unigram.most_common():
    print(f"{word:<15}{freq}")


# --------------------------------------------------
# FUNCTION TO CREATE N-GRAMS
# --------------------------------------------------

def make_ngrams(tokens, n):

    result = []

    for i in range(len(tokens) - n + 1):
        result.append(tuple(tokens[i:i+n]))

    return result


# --------------------------------------------------
# 4. BIGRAM
# --------------------------------------------------

bigram = make_ngrams(tokens, 2)
bigram_freq = Counter(bigram)

print("\n========== BIGRAM FREQUENCY TABLE ==========")
print("Bigram\t\t\tFrequency")
print("---------------------------------------------")

for bg, freq in bigram_freq.most_common():
    print(f"{' '.join(bg):<25}{freq}")


# --------------------------------------------------
# 5. TRIGRAM
# --------------------------------------------------

trigram = make_ngrams(tokens, 3)
trigram_freq = Counter(trigram)

print("\n========== TRIGRAM FREQUENCY TABLE ==========")
print("Trigram\t\t\t\tFrequency")
print("----------------------------------------------")

for tg, freq in trigram_freq.most_common():
    print(f"{' '.join(tg):<30}{freq}")


# --------------------------------------------------
# 6. FOUR-GRAM
# --------------------------------------------------

fourgram = make_ngrams(tokens, 4)
fourgram_freq = Counter(fourgram)

print("\n========== FOUR-GRAM FREQUENCY TABLE ==========")
print("Four-gram\t\t\t\tFrequency")
print("------------------------------------------------")

for fg, freq in fourgram_freq.most_common():
    print(f"{' '.join(fg):<35}{freq}")


# --------------------------------------------------
# 7. NEXT WORD PREDICTION
# --------------------------------------------------

def predict_next_word(context, ngram_freq):

    context = tuple(context.lower().split())

    candidates = {}

    for gram, freq in ngram_freq.items():

        if gram[:-1] == context:
            candidates[gram[-1]] = freq

    if candidates:
        return max(candidates, key=candidates.get)

    return "No prediction found"


# --------------------------------------------------
# 8. NEXT WORD PREDICTION
# --------------------------------------------------

print("\n========== NEXT WORD PREDICTION ==========")


# BIGRAM
prediction = predict_next_word(
    "like",
    bigram_freq
)

print("\nBigram:")
print("Input : I like")
print("Predicted next word :", prediction)


# TRIGRAM
prediction = predict_next_word(
    "i like",
    trigram_freq
)

print("\nTrigram:")
print("Input : I like")
print("Predicted next word :", prediction)


# FOUR-GRAM
prediction = predict_next_word(
    "i like to",
    fourgram_freq
)

print("\nFour-gram:")
print("Input : I like to")
print("Predicted next word :", prediction)
```
