```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

PART A - NLTK

Hindi Text

```python
hindi_text = """
प्राकृतिक भाषा प्रसंस्करण कृत्रिम बुद्धिमत्ता का एक महत्वपूर्ण क्षेत्र है।
यह कंप्यूटर को मानव भाषा को समझने, संसाधित करने और उसका विश्लेषण करने में सक्षम बनाता है।
आज प्राकृतिक भाषा प्रसंस्करण का उपयोग शिक्षा, स्वास्थ्य, व्यवसाय और संचार जैसे विभिन्न क्षेत्रों में किया जाता है।
विद्यार्थी हिंदी भाषा के वाक्यों, शब्दों और उनके अर्थों का अध्ययन करके भाषा की संरचना को समझ सकते हैं।"""
```

Sentence Detection

```python
sentences = nltk.tokenize.RegexpTokenizer(r'[^।]+').tokenize(hindi_text)

print("===== SENTENCE DETECTION =====")

for i, sentence in enumerate(sentences, 1):
    print(f"Sentence {i}: {sentence.strip()}")
```

Tokens

```python
import re

tokens = re.findall(r'[\u0900-\u097F]+', hindi_text)

print(f"Tokens ({len(tokens)}):")
print(tokens)
```

Statistical Analysis over Tokens

```python
from collections import Counter

total_tokens = len(tokens)
unique_tokens = len(set(tokens))

avg_token_length = sum(len(token) for token in tokens) / total_tokens

frequency = Counter(tokens)

print("===== STATISTICAL ANALYSIS OVER TOKENS =====")

print("Total Tokens:", total_tokens)
print("Unique Tokens:", unique_tokens)
print("Average Token Length:", round(avg_token_length, 2))

print("\nTop 5 Tokens:")
for token, count in frequency.most_common(5):
    print(token, ":", count)
```

# *Lemmatization

```python
# NLTK does not have a Hindi lemmatizer, so don't use WordNetLemmatizer for Hindi.

# For Hindi, we'll do lemmatization using Stanza
```

Stop-word Removal

```python
hindi_stopwords = {
    'और', 'का', 'के', 'की', 'को', 'से', 'में', 'पर', 'यह', 'वह',
    'है', 'हैं', 'था', 'थे', 'थी', 'एक', 'जो', 'कि', 'तो', 'ही',
    'भी', 'ने', 'इस', 'उस', 'इन', 'उन', 'मैं', 'हम', 'आप',
    'मेरा', 'मेरी', 'मेरे', 'अपना', 'अपनी', 'अपने', 'लिए', 'तक',
    'जब', 'जहाँ', 'क्यों', 'क्या', 'कैसे', 'बहुत', 'अब', 'नहीं' ,'है।'
}

# Stop-word removal
filtered_tokens = [
    token for token in tokens
    if token not in hindi_stopwords
]

# Display result with count
print(f"After stop-word removal ({len(filtered_tokens)}):")
print(filtered_tokens)
```

PART B - STANZA


```python
!pip install stanza
import stanza

stanza.download('hi')
```

```python
nlp_hi = stanza.Pipeline(
    lang='hi',
    processors='tokenize,pos,lemma'
)
```

Sentence Detection

```python
doc = nlp_hi(hindi_text)

print("===== SENTENCE DETECTION =====")

for i, sentence in enumerate(doc.sentences, 1):
    print(f"Sentence {i}: {sentence.text}")
```

Tokens

```python
tokens_stanza = []

for sentence in doc.sentences:
    for word in sentence.words:
        tokens_stanza.append(word.text)

print(f"Tokens ({len(tokens_stanza)}):")
print(tokens_stanza)
```

Statistical Analysis over Tokens

```python
from collections import Counter

total_tokens = len(tokens_stanza)
unique_tokens = len(set(tokens_stanza))

avg_token_length = sum(
    len(token) for token in tokens_stanza
) / total_tokens

frequency = Counter(tokens_stanza)

print("===== STATISTICAL ANALYSIS OVER TOKENS =====")

print("Total Tokens:", total_tokens)
print("Unique Tokens:", unique_tokens)
print("Average Token Length:", round(avg_token_length, 2))

print("\nTop 5 Tokens:")
for token, count in frequency.most_common(5):
    print(token, ":", count)
```

Lemmatization

```python
print("===== LEMMAS =====")

for sentence in doc.sentences:
    for word in sentence.words:
        print(word.text, "->", word.lemma)
```
