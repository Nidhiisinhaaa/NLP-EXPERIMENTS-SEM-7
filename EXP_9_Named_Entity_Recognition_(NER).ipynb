````python
import nltk

# Download required NLTK resources
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('maxent_ne_chunker')
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')

# Input sentence
text = """
On 26 January 2027 at 9:00 AM, Priya Mehta from Google purchased
3 Samsung Galaxy S26 phones in New Delhi, India, for ₹1,80,000
with a 15% discount during the Republic Day celebration.
"""

# Tokenization
tokens = nltk.word_tokenize(text)

# POS Tagging
pos_tags = nltk.pos_tag(tokens)

# Named Entity Recognition
named_entities = nltk.ne_chunk(pos_tags)

# Display complete NER tree
print("Named Entity Recognition using NLTK:\n")
print(named_entities)

# Extract and display classified entities
print("\nClassified Entities:\n")

for chunk in named_entities:
    if hasattr(chunk, 'label'):
        entity = " ".join(word for word, tag in chunk)
        print(f"{entity} -> {chunk.label_ if hasattr(chunk, 'label_') else chunk.label()}")
````

spaCy

````python
import spacy

# Load the English spaCy model
nlp = spacy.load("en_core_web_sm")

# Input sentence
text = """
On 26 January 2027 at 9:00 AM, Priya Mehta from Google purchased
3 Samsung Galaxy S26 phones in New Delhi, India, for ₹1,80,000
with a 15% discount during the Republic Day celebration.
"""

# Process the text
doc = nlp(text)

# Display named entities
print("Named Entity Recognition using spaCy:\n")

for ent in doc.ents:
    print(f"{ent.text} -> {ent.label_} -> {spacy.explain(ent.label_)}")
````

Stanza

````python
# Install Stanza
!pip install -q stanza

import stanza

# Download English model
stanza.download("en")

# Create NLP pipeline
nlp = stanza.Pipeline(
    lang="en",
    processors="tokenize,ner"
)

# Input sentence
text = """
On 26 January 2027 at 9:00 AM, Priya Mehta from Google purchased
3 Samsung Galaxy S26 phones in New Delhi, India, for ₹1,80,000
with a 15% discount during the Republic Day celebration.
"""

# Process text
doc = nlp(text)

print("Named Entity Recognition using Stanza:\n")

# Display named entities
for ent in doc.ents:
    print(f"{ent.text} -> {ent.type}")
````

**Among NLTK, spaCy, and Stanford NER, spaCy is more suitable for practical Named Entity Recognition because it provides a pretrained NER model, supports a wider range of entity types, is easy to integrate with Python, and offers efficient processing. NLTK is useful for learning and basic NLP tasks, while Stanford NER is useful for traditional NER and research-oriented applications.**

````python
import stanza

# Create Hindi NLP pipeline
nlp = stanza.Pipeline(
    lang="hi",
    processors="tokenize,ner"
)

# Hindi input sentence
text = """
26 जनवरी 2027 को सुबह 9 बजे, प्रिय शर्मा ने दिल्ली में
Google से ₹1,80,000 में 3 Samsung Galaxy S26 फोन खरीदे।
"""

# Process the text
doc = nlp(text)

print("Named Entity Recognition using Stanza:\n")

# Print named entities
for ent in doc.ents:
    print(f"{ent.text} -> {ent.type}")
````

The Indian-language NER tagset defines NEP = Person, NEL = Location, NEO = Organization, NEN = Number, and NETI = Time.
