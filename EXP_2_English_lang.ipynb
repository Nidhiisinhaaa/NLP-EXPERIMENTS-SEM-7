**1. NLTK**

```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from collections import Counter
```

Sample English text

```python
text = """Machine learning is an important part of artificial intelligence.
It allows computers to learn from data and make predictions.Students are studying machine learning algorithms.
These algorithms are used in healthcare, education, and business.
Modern applications use machine learning to solve complex problems."""
```

1.1 Sentence Detection using NLTK

```python
sentences = sent_tokenize(text)

print("NLTK - Sentence Detection\n")

for i, sentence in enumerate(sentences, 1):
    print(f"Sentence {i}: {sentence}")
```

2. Tokenization using NLTK

```python
tokens = word_tokenize(text)

print("NLTK - Tokenization\n")
print(f"Tokens ({len(tokens)}):", tokens)
```

3. Statistical Analysis over Tokens

```python
from collections import Counter

print("--- Statistical analysis over tokens ---")

print(f"Total tokens    : {len(tokens)}")
print(f"Unique tokens   : {len(set(tokens))}")

avg_token_length = sum(len(token) for token in tokens) / len(tokens)
print(f"Avg token length: {avg_token_length:.1f}")

frequency = Counter(tokens)
print(f"Top 5 frequent : {frequency.most_common(5)}")
```

Stop Word Removal using NLTK

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

filtered_words = [
    word for word in tokens
    if word.lower() not in stop_words
]

print("NLTK - Stop Word Removal\n")
print(f"Stop Word Removal ({len(filtered_words)}):", filtered_words)
```

Lemmatization using NLTK

```python
from nltk.stem import WordNetLemmatizer

# Create lemmatizer
lemmatizer = WordNetLemmatizer()

print("NLTK - Lemmatization\n")
print("Word -> Lemma:")

for word in filtered_words:
    if word.isalpha():          # removes punctuation like . , !
        lemma = lemmatizer.lemmatize(word)
        print(f"{word} -> {lemma}")
```

```python
import spacy

text = """Machine learning is an important part of artificial intelligence.
It allows computers to learn from data and make predictions.
Students are studying machine learning algorithms.
These algorithms are used in healthcare, education, and business.
Modern applications use machine learning to solve complex problems"""

nlp = spacy.load("en_core_web_sm")
doc = nlp(text)
```

1. Sentence Detection

```python
print("===== spaCy - Sentence Detection =====\n")

for i, sentence in enumerate(doc.sents, 1):
    print(f"Sentence {i}: {sentence.text}")
```

2. Tokenization

```python
tokens = [token.text for token in doc]

print(tokens)
```

3. Statistical Analysis over Tokens

```python
from collections import Counter

words = [
    token.text
    for token in doc
    if token.is_alpha
]

frequency = Counter(words)

print("--- Statistical analysis over tokens ---")

print(f"Total tokens    : {len(words)}")
print(f"Unique tokens   : {len(set(words))}")

avg_token_length = sum(len(word) for word in words) / len(words)
print(f"Avg token length: {avg_token_length:.1f}")

print(f"Top 5 frequent : {frequency.most_common(5)}")
```

Stop Word Removal

```python
print("\n===== spaCy - Stop Word Removal =====\n")

filtered_words = [
    token.text
    for token in doc
    if token.is_alpha and not token.is_stop
]

print(filtered_words)
```

Lemmatization

```python
print("\n===== spaCy - Lemmatization =====\n")

print("Word -> Lemma:")

for token in doc:
    if token.is_alpha and not token.is_stop:
        print(f"{token.text} -> {token.lemma_}")
```

3. Stanza is the Stanford NLP

1. Load Stanza and English model

```python
import stanza

# Load Stanza English pipeline
nlp_stanza = stanza.Pipeline(
    'en',
    processors='tokenize,pos,lemma'
)
```

```python
# Input text
text = """The environment provides essential resources for all living organisms.
Clean air, fresh water, and healthy soil are necessary for human and animal life.
Deforestation and pollution are increasing environmental problems around the world.
People can protect nature by planting trees, reducing waste, and saving water.
Sustainable development can help create a cleaner and healthier planet for future generations."""

# Process the text
doc_stanza = nlp_stanza(text)
```

Sentence Detection

```python
print("===== Stanza - Sentence Detection =====\n")

for i, sentence in enumerate(doc_stanza.sentences, 1):
    print(f"Sentence {i}: {sentence.text}")
```

2. Tokenization

```python
print("\nStanza - Tokenization\n")

tokens = []

for sentence in doc_stanza.sentences:
    for word in sentence.words:
        tokens.append(word.text)

print(f"Tokens ({len(tokens)}): {tokens}")
```

3. Statistical Analysis

```python
from collections import Counter

# Get tokens
tokens = [
    word.text
    for sentence in doc_stanza.sentences
    for word in sentence.words
    if word.text.isalpha()
]

# Frequency calculation
frequency = Counter(tokens)

# Average token length
avg_length = sum(len(word) for word in tokens) / len(tokens)

print("\n--- Statistical analysis over tokens ---")

print(f"Total tokens    : {len(tokens)}")
print(f"Unique tokens   : {len(set(tokens))}")
print(f"Avg token length: {avg_length:.1f}")
print(f"Top 5 frequent  : {frequency.most_common(5)}")
```

Stop Word Removal

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

filtered_words = [
    word for word in tokens
    if word.lower() not in stop_words
]

print(f"stop-word removal ({len(filtered_words)}): {filtered_words}")
```

Lemmatization

```python
print("\nStanza - Lemmatization\n")

print("Word -> Lemma:")

for sentence in doc_stanza.sentences:
    for word in sentence.words:
        if word.text.isalpha():
            print(f"{word.text} -> {word.lemma}")
```

```python
import nltk
from collections import Counter
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# Download required NLTK data
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")
nltk.download("wordnet")

# Input text
text = """Machine learning is transforming modern education and healthcare.
Researchers use data to discover useful patterns and improve services.
Natural language processing enables computers to understand human communication."""

# 1. Sentence Detection
sentences = sent_tokenize(text)

# 2. Tokenization
tokens = word_tokenize(text)
words = [w.lower() for w in tokens if w.isalpha()]

print(">>> Colab Output\n")

print("1. SENTENCE DETECTION")
print("Total Sentences:", len(sentences))

for i, s in enumerate(sentences, 1):
    print(f"{i}. {s}")

# 3. Statistical Analysis
freq = Counter(words)

print("\n2. TOKENIZATION")
print(words)

print("\n3. STATISTICAL ANALYSIS")
print("Total Tokens:", len(words))
print("Unique Tokens:", len(set(words)))

print("Word Frequency:")
for word, count in freq.most_common():
    print(f"{word} : {count}")

# 4. Lemmatization
print("\n4. LEMMATIZATION")

lemmatizer = WordNetLemmatizer()

for w in words:
    print(w, "->", lemmatizer.lemmatize(w))

# 5. Stop Word Removal
stop_words = set(stopwords.words("english"))
filtered = [w for w in words if w not in stop_words]

removed = [w for w in words if w in stop_words]

print("\n5. STOP WORD REMOVAL")
print("Removed:", removed)
print("Filtered:", filtered)
```
